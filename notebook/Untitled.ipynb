```python
import pandas as pd

df = pd.read_csv("../data/flights2024.csv", low_memory=False)

# Ensure FL_DATE is a date
df["FL_DATE"] = pd.to_datetime(df["FL_DATE"], errors="coerce").dt.date

# Rename ONLY what the old PBIX expects
df = df.rename(columns={
    "YEAR": "Year",
    "MONTH": "Month Number",
    "DAY_OF_MONTH": "Day",
    "DAY_OF_WEEK": "Day Of Week"
})

# Drop duplicate delay columns (PBIX never uses these)
df = df.drop(columns=["DEP_DELAY_NEW", "ARR_DELAY_NEW"], errors="ignore")

# Save Power BI–ready sample
df.to_csv("dec24_for_powerbi.csv", index=False)

print("✅ dec24_for_powerbi.csv created")
print("Key columns now:")
print(df[["FL_DATE", "Year", "Month Number", "Day", "Day Of Week"]].head())
```
